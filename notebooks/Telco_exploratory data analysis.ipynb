# adding features

```python
import pandas as pd
import numpy as np
```

```python
df= pd.read_csv("Telco-Customer-Churn_cleaned.csv")
```

```python
df.shape
```

```python
#chargeCategory
df["ChargeCategory"] = pd.cut(
    df["MonthlyCharges"],
    bins=[0,40,80,150],
    labels=["Low","Medium","High"]
)
```

```python
#charge category nto numeric columns
df = pd.get_dummies(df, columns=["ChargeCategory"], drop_first=True)
```

```python
bool_cols = df.select_dtypes(include='bool').columns

for col in bool_cols:
    df[col] = df[col].astype(int)

df.head()
```

```python
df.to_csv("add_feature_monthly charges category.csv",index=False)
```
